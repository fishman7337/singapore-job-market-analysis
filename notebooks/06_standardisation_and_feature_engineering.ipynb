<h3>2.11.1 Feaure Engineering For Dataset 1</h3>

<p>For Dataset 1, we will be creating 2 new Features which are 'Field of Study' and 'uni_short'. These will be important when we carry out the Data Analysis and Visualisation as we are able to see the a more streamlined categorical data. This will be done through mapping.</p>

```python
# ---------- Mapping Of Field Of Studies ---------- #
degree_categories = {
    "Business": ["business", "accountancy", "finance", "commerce", "management", "marketing", "economics"],
    "Engineering": ["engineering", "mechanical", "electrical", "civil", "aerospace", "chemical", "industrial"],
    "Science": ["science", "biology", "chemistry", "physics", "biomedical", "mathematics"],
    "Computer Science & IT": ["computer", "computing", "data", "artificial intelligence", "cybersecurity", "software", "information technology"],
    "Humanities & Social Sciences": ["arts", "history", "philosophy", "sociology", "psychology", "political science"],
    "Healthcare & Medicine": ["medicine", "nursing", "pharmacy", "dental", "health", "nutrition"],
    "Law": ["law", "legal", "jurisprudence"],
    "Education": ["education", "teaching", "pedagogy"],
    "Others": []
}

# ---------- Function For Classification ---------- #
def categorize_degree(degree):
    degree_lower = str(degree).lower()
    for category, keywords in degree_categories.items():
        if any(keyword in degree_lower for keyword in keywords):
            return category
    return "Others"  # Default if no match

# ---------- Applying Of Classification ---------- #
df1["Field of Study"] = df1["degree"].apply(categorize_degree)

# ---------- Mapping Of University Short Form ---------- #
university_map = {
    "Nanyang Technological University": "NTU",
    "National University of Singapore": "NUS",
    "Singapore Management University": "SMU",
    "Singapore Institute of Technology": "SIT",
    "Singapore University of Social Sciences": "SUSS",
    "Singapore University of Technology and Design": "SUTD"
}

# ---------- Applying Of Mapping ---------- #
df1['uni_short'] = df1['university'].map(university_map).fillna(df1['university'])

# ---------- Displaying DataFrame ---------- #
df1.head(5).style.background_gradient(cmap="Blues")
```

<p>With reference to the output of the DataFrame, we are able to see that the new Data Column has been created.</p>
